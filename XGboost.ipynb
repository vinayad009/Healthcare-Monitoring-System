```python
# Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report
from xgboost import XGBClassifier

# Load the dataset
file_path = "cardio_train.csv"  # Replace with your file path
data = pd.read_csv(file_path, delimiter=";")

# Clean and preprocess the data
# Remove outliers based on reasonable ranges
valid_height_range = (100, 250)  # Height in cm
valid_weight_range = (30, 200)  # Weight in kg
valid_ap_hi_range = (90, 200)   # Systolic BP
valid_ap_lo_range = (60, 120)   # Diastolic BP

data = data[
    (data['height'].between(*valid_height_range)) &
    (data['weight'].between(*valid_weight_range)) &
    (data['ap_hi'].between(*valid_ap_hi_range)) &
    (data['ap_lo'].between(*valid_ap_lo_range))
]

# Convert age from days to years
data['age'] = (data['age'] / 365).astype(int)

# Define the target and features
target = 'cardio'
features = ['age', 'gender', 'height', 'weight', 'ap_hi', 'ap_lo', 'active']

X = data[features]
y = data[target]

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)

# Train an XGBoost classifier
# model = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
# model.fit(X_train, y_train)

model = XGBClassifier(eval_metric='logloss', random_state=0)
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)

# Display the results
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("Classification Report:\n", classification_rep)
```

```python
# Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report
from xgboost import XGBClassifier

# Load the dataset
file_path = "cardio_train.csv"  # Replace with your file path
data = pd.read_csv(file_path, delimiter=";")

# Clean and preprocess the data
# Remove outliers based on reasonable ranges
valid_height_range = (100, 250)  # Height in cm
valid_weight_range = (30, 200)  # Weight in kg
valid_ap_hi_range = (90, 200)   # Systolic BP
valid_ap_lo_range = (60, 120)   # Diastolic BP

data = data[
    (data['height'].between(*valid_height_range)) &
    (data['weight'].between(*valid_weight_range)) &
    (data['ap_hi'].between(*valid_ap_hi_range)) &
    (data['ap_lo'].between(*valid_ap_lo_range))
]

# Convert age from days to years
data['age'] = (data['age'] / 365).astype(int)

# Define the target and all features
target = 'cardio'
features = ['age', 'gender', 'height', 'weight', 'ap_hi', 'ap_lo', 
            'cholesterol', 'gluc', 'smoke', 'alco', 'active']

X = data[features]
y = data[target]

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)

# Train an XGBoost classifier
model = XGBClassifier(eval_metric='logloss', random_state=0)
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)

# Display the results
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("Classification Report:\n", classification_rep)

# Function to take manual input for prediction
def get_user_input(features):
    print("\nProvide the following inputs for prediction:")
    input_data = []
    for feature in features:
        value = float(input(f"Enter {feature}: "))
        input_data.append(value)
    return pd.DataFrame([input_data], columns=features)

# Take user input and predict
user_input = get_user_input(features)
prediction = model.predict(user_input)[0]

# Display prediction result
print("\nPrediction Result:")
if prediction == 1:
    print("The model predicts: Cardiovascular Disease")
else:
    print("The model predicts: No Cardiovascular Disease")
```
